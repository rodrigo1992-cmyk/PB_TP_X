```python
def pre_tratamento_pontuacao(descricao_vaga):
    descricao_vaga = descricao_vaga.replace(";", ", ")
    descricao_vaga = descricao_vaga.replace("(", ", ")
    descricao_vaga = descricao_vaga.replace("),", ", ")
    descricao_vaga = descricao_vaga.replace(")", ", ")
    descricao_vaga = descricao_vaga.replace("/", ", ")
    descricao_vaga = descricao_vaga.replace(".", ". ")
    descricao_vaga = descricao_vaga.replace("|", ", ")
    descricao_vaga = descricao_vaga.replace(",", ", ")
    descricao_vaga = descricao_vaga.replace(",,", ", ")
    descricao_vaga = descricao_vaga.replace("  ", " ")
    descricao_vaga = descricao_vaga.replace(",, ", ", ")
    descricao_vaga = descricao_vaga.replace(", .", ".")
    descricao_vaga = descricao_vaga.replace(". ,", ".")

    return descricao_vaga

def tratar_lista(lst_resp):
    #splitar a lista de software
    lst_sw = [item.replace(" and ", ",").replace(" e ", ",").replace(", ", ",") for item in lst_resp]
    lst_sw = [item.split(",") for item in lst_sw]

    #trasnformar a lista em um conjunto
    set_sw = set()
    for item in lst_sw:
        set_sw.update(item)
    
    return set_sw
```

```python
descricao_vaga_portugues = []
descricao_vaga_ingles = []


for i in descricao_vaga_portugues_orig:
    descricao_vaga_portugues.append(pre_tratamento_pontuacao(i))

for i in descricao_vaga_ingles_orig:
    descricao_vaga_ingles.append(pre_tratamento_pontuacao(i))
```

# Teste 1 - Modelo Tiny Roberta - Em Inglês

#### O modelo de QA Tiny Roberta só retorna a primeira sequência dentro do texto. Ou seja, se os softwares forem citados em frases distintas o modelo só irá retornar a primeira ocorrência.

```python
from transformers import pipeline

pipe = pipeline("question-answering", model="deepset/tinyroberta-squad2")

prompt = "Which software names are mentioned in the text?"
lst_resp = []

for vaga in descricao_vaga_ingles:
    resposta = pipe(question = prompt, context = vaga)
    lst_resp.append(resposta['answer'])

set_sw_TinyBerta = tratar_lista(lst_resp)
set_sw_TinyBerta
```

# Teste 3 - Modelo Bert - Fining Tuning em Português
### Abordagem passando uma frase por vez
#### Nesta abordagem, mesmo que não exista uma resposta na frase o modelo retorna algo, trazendo uma grande quantidade de falsos positivos. Para melhorar, fiz um filtro no Score baixo, removendo boa parte dos falsos positivos. 

```python
from transformers import pipeline
import pandas as pd

pipe = pipeline("question-answering", model="pierreguillou/bert-base-cased-squad-v1.1-portuguese")

prompt = "Quais Softwares são mencionados no texto?"
lst_resp = []

for vaga in descricao_vaga_portugues:
    lista_frases = vaga.split(".")
    for frase in lista_frases:
        if frase:
            resposta = pipe(question = prompt, context = frase)
            lst_resp.append(resposta)

#converter o valor de científico para float
df = pd.DataFrame(lst_resp)

#arredondar o score e filtrar o score
df['score'] = df['score'].apply(lambda x: round(x, 6))
df.sort_values(by='score', ascending=False, inplace=True)
df
```

```python
#Filtrando o score
df = df[df['score'] > 0.001]

#transformar a coluna answer em uma lista
lista_software = df['answer'].to_list()

set_sw_BertPT_V2 = tratar_lista(lista_software)

set_sw_BertPT_V2
```

# Comparando os Resultados

```python
import regex as re

def tratar_respostas(item):
    item = item.lower()
    item = re.sub(r'^e\s+', '', item)
    item = re.sub(r'\s+e\s+', '', item)
    item = re.sub(r'\s+e$', '', item)
    item = re.sub(r'^ou\s+', '', item)
    item = re.sub(r'\s+ou\s+', '', item)
    item = re.sub(r'\s+ou$', '', item)

    return item

#juntar todos os conjuntos
set_sw = set_sw_TinyBerta.union(set_sw_BertPT).union(set_sw_BertPT_V2).union(set_sw_SpacyEN).union(set_sw_SpacyPT)

#converter todos os valores para minúsculo
set_sw = {item.lower() for item in set_sw}

#converter o conjunto em uma lista
lst_sw_completa = list(set_sw)
lst_sw_completa = [tratar_respostas(item) for item in lst_sw_completa]
```

```python
df = pd.DataFrame(columns=['Termos','conjunto_verdade','TinyBerta', 'BertPT', 'BertPT_V2', 'SpacyEN', 'SpacyPT'])

df['Termos'] = lst_sw_completa

def checar_resultados(df, conj):
    conj_tratado = [tratar_respostas(item) for item in conj]
    avaliacao = []
    for termo in df['Termos']:
        if termo in conj_tratado:
            avaliacao.append(1)
        else: 
            avaliacao.append(0)
    return avaliacao

df['conjunto_verdade'] = checar_resultados(df, set_verdade)
df['TinyBerta'] = checar_resultados(df, set_sw_TinyBerta)
df['BertPT'] = checar_resultados(df, set_sw_BertPT)
df['BertPT_V2'] = checar_resultados(df, set_sw_BertPT_V2)
df['SpacyEN'] = checar_resultados(df, set_sw_SpacyEN)
df['SpacyPT'] = checar_resultados(df, set_sw_SpacyPT)

df
```

```python
#calcular a acurácia
def calcular_acertos(df, coluna_verdade, coluna_calcular):
    total_verdade = df[df[coluna_verdade] == 1].shape[0]
    total_respostas = df[df[coluna_calcular] == 1].shape[0]

    acertos = df[(df[coluna_verdade] == 1) & (df[coluna_calcular] == 1)].shape[0]

    perc_acerto = int(round((acertos / total_respostas) * 100, 0))

    erros = df[(df[coluna_verdade] == 0) & (df[coluna_calcular] == 1)].shape[0]
    perc_erros = int(round((erros / total_respostas) * 100 ,0))

    nao_preditos = df[(df[coluna_verdade] == 1) & (df[coluna_calcular] == 0)].shape[0]
    perc_nao_preditos = int(round((nao_preditos / total_verdade) * 100, 0))

    acuracia = int(round((acertos / total_verdade) * 100, 0))

    return total_respostas, acuracia, perc_acerto, perc_erros, perc_nao_preditos

Scores_TinyBerta = calcular_acertos(df, 'conjunto_verdade', 'TinyBerta')
Scores_BertPT = calcular_acertos(df, 'conjunto_verdade', 'BertPT')
Scores_BertPT_V2 = calcular_acertos(df, 'conjunto_verdade', 'BertPT_V2')
Scores_SpacyEN = calcular_acertos(df, 'conjunto_verdade', 'SpacyEN')
Scores_SpacyPT = calcular_acertos(df, 'conjunto_verdade', 'SpacyPT')

#Adicionar os valores a um novo dataframe
df_scores = pd.DataFrame(columns=['Modelo', 'Total Itens', 'acuracia %', '% Acertos','% Erros', 'Não Preditos'])

df_scores.loc[0] = ['TinyBerta'] + list(Scores_TinyBerta)
df_scores.loc[1] = ['BertPT'] + list(Scores_BertPT)
df_scores.loc[2] = ['BertPT_V2'] + list(Scores_BertPT_V2)
df_scores.loc[3] = ['SpacyEN'] + list(Scores_SpacyEN)
df_scores.loc[4] = ['SpacyPT'] + list(Scores_SpacyPT)
df_scores
```
