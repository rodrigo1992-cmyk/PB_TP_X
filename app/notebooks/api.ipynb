Através do terminal, andar até a pasta onde está o arquivo da API

uvicorn main:appApi --reload


1. O arquivo Main:
   1. Instancia o FastAPI
   2. inclui import as rotas do arquivo routs na instância 
   3. inicia o servidor rodando a instância do FastAPI
   
2. O arquivo routes:
   1. Cada rota corresponde a uma função que retorna um determinado valor
   
3. O arquivo do Streamlit
   1. Faz um get ao IP e recebe o valor



### No arquivo main do backend deve ter:
app = FastAPI()

app.include_router(router)

if __name__ == "__main__":
    import uvicorn
    uvicorn.run(app, host="0.0.0.0", port=8000)

### No arquivo router deve ter a instrução:
router = APIRouter()

@router.get("/data")
async def read_data():
    return {"message": "Hello from FastAPI!"}


### No arquivo Streamlit deve ter a instrução:
response = requests.get("http://localhost:8000/data")
data = response.json()
